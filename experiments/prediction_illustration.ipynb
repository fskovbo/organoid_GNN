# Prediction Illustration

Train GIN models at depths 0, 2, and 4 on the original marker channels and on a matched LGR5-only input variant where every non-LGR5 marker channel is zeroed. The final figure projects predicted curvature for a single exemplar organoid onto its mesh, using a shared color scale across all panels.

**Helpers**

```python
def safe_filename(name):
    text = str(name).strip().replace("/", "_")
    cleaned = "".join(
        char if (char.isalnum() or char in "._-") else "_" for char in text
    ).strip("._-")
    while "__" in cleaned:
        cleaned = cleaned.replace("__", "_")
    return cleaned or "figure"


def save_plotly_figure(fig, name, *, scale=2):
    stem = safe_filename(name)
    html_path = FIGURES_DIR / f"{stem}.html"
    fig.write_html(str(html_path), include_plotlyjs="cdn")
    png_path = FIGURES_DIR / f"{stem}.png"
    try:
        fig.write_image(str(png_path), scale=scale)
    except Exception as exc:
        print(f"Plotly PNG export skipped for {stem}: {exc}")
    return html_path


def jsonable(obj):
    if isinstance(obj, dict):
        return {str(key): jsonable(value) for key, value in obj.items()}
    if isinstance(obj, (list, tuple, set)):
        return [jsonable(value) for value in obj]
    if isinstance(obj, Path):
        return str(obj)
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    if isinstance(obj, np.generic):
        return obj.item()
    if isinstance(obj, torch.Tensor):
        return obj.detach().cpu().tolist()
    if isinstance(obj, (torch.dtype, np.dtype)):
        return str(obj)
    return obj


def select_target_column(values, target_index=0):
    array = np.asarray(values)
    if array.ndim == 1:
        return array.reshape(-1)
    if array.ndim == 2 and array.shape[1] == 1:
        return array[:, 0]
    if array.ndim == 2:
        return array[:, int(target_index)]
    raise ValueError(f"Expected 1-D or 2-D values, got {array.shape}.")


def graph_slice_bounds(graphs_in, graph_index):
    sizes = [int(graph.x.shape[0]) for graph in graphs_in]
    start = sum(sizes[:int(graph_index)])
    return start, start + sizes[int(graph_index)]


def marker_index(name, marker_names_in):
    names = list(marker_names_in)
    if name in names:
        return names.index(name)
    lower = {str(marker).lower(): index for index, marker in enumerate(names)}
    key = str(name).lower()
    if key not in lower:
        raise ValueError(f"Marker {name!r} not found in {names}.")
    return lower[key]


def zero_marker_channels(graphs_in, markers_to_remove, marker_names_in):
    marker_indices = [
        marker_index(marker, marker_names_in) for marker in markers_to_remove
    ]
    graphs_out = [copy.deepcopy(graph) for graph in graphs_in]
    counts = {marker: 0 for marker in markers_to_remove}
    for graph in graphs_out:
        graph.x = graph.x.clone()
        for marker, marker_idx in zip(markers_to_remove, marker_indices):
            counts[marker] += int((graph.x[:, marker_idx] > 0.5).sum().item())
            graph.x[:, marker_idx] = 0
    return graphs_out, counts


def all_except_markers(marker_names_in, keep_markers):
    keep_indices = {
        marker_index(marker, marker_names_in) for marker in keep_markers
    }
    return [
        marker
        for index, marker in enumerate(marker_names_in)
        if index not in keep_indices
    ]
```

**Split And Transform Targets**

```python
from src.data.metadata import (
    add_log_metadata_features,
    infer_global_dim,
    promote_metadata_to_graph_tensors,
    snapshot_graph_metadata,
    strip_graph_metadata,
)
from src.data.splits import graph_metadata_key, train_val_split_graphs
from src.data.target_transforms import (
    AsinhStandardizeTransform,
    standardize_graph_global_features,
)

FIELD_SPECS = [{
    "meta_keys": [
        "log_surface_area",
        "log_volume",
        "log_volume_over_area",
        "log_num_cells",
    ],
    "attr_name": "global_feat",
    "kind": "graph_vector",
    "dtype": torch.float32,
}]

if USE_GLOBAL_FEATURES:
    graphs = promote_metadata_to_graph_tensors(
        add_log_metadata_features(graphs, inplace=False),
        FIELD_SPECS,
        inplace=False,
    )
else:
    raise ValueError("This notebook requires USE_GLOBAL_FEATURES=True.")

g_train_raw, g_val_raw, split_info = train_val_split_graphs(
    graphs,
    val_frac=VAL_FRAC,
    seed=SPLIT_SEED,
    force_val_keys=FORCED_VAL_KEYS,
    key_fn=graph_metadata_key,
    inplace=False,
)
print(f"Split -> train: {len(g_train_raw)} | val: {len(g_val_raw)}")

train_meta_lookup = snapshot_graph_metadata(g_train_raw)
val_meta_lookup = snapshot_graph_metadata(g_val_raw)
g_train = strip_graph_metadata(g_train_raw, inplace=False)
g_val = strip_graph_metadata(g_val_raw, inplace=False)

global_center, global_scale = standardize_graph_global_features(
    g_train,
    g_val,
    attr_name="global_feat",
    robust=False,
)
target_transform = AsinhStandardizeTransform(robust=True)
target_transform.fit(g_train)
target_transform.transform_graphs(g_train, in_place=True)
target_transform.transform_graphs(g_val, in_place=True)

g_train_reduced, train_zero_counts = zero_marker_channels(
    g_train, markers_removed_for_reduced, marker_names
)
g_val_reduced, val_zero_counts = zero_marker_channels(
    g_val, markers_removed_for_reduced, marker_names
)

split_summary_df = pd.DataFrame([{
    "n_graphs_filtered": len(graphs),
    "n_train_graphs": len(g_train),
    "n_val_graphs": len(g_val),
    "n_train_nodes": sum(int(graph.x.shape[0]) for graph in g_train),
    "n_val_nodes": sum(int(graph.x.shape[0]) for graph in g_val),
    "val_frac": VAL_FRAC,
    "split_seed": SPLIT_SEED,
    "forced_val_keys": repr(sorted(FORCED_VAL_KEYS)),
}])
zero_summary_df = pd.DataFrame([
    {
        "variant": REDUCED_VARIANT_KEY,
        "marker": marker,
        "train_positive_values_zeroed": int(train_zero_counts[marker]),
        "val_positive_values_zeroed": int(val_zero_counts[marker]),
    }
    for marker in markers_removed_for_reduced
])
split_summary_df.to_csv(TABLES_DIR / "split_summary.csv", index=False)
zero_summary_df.to_csv(TABLES_DIR / "reduced_variant_zero_counts.csv", index=False)
display(split_summary_df)
display(zero_summary_df)
```

**Train Models**

```python
from src.models.gnn import GINCurvature
from src.training.loop import TrainConfig, train
from src.training.losses import WeightedLossTerm, edge_loss_term


def infer_target_dim(graphs_in):
    y = graphs_in[0].y
    return 1 if y.ndim == 1 else int(y.shape[1])


def make_gin_model(graphs_in, depth):
    return GINCurvature(
        n_markers=int(graphs_in[0].x.size(1)),
        global_dim=infer_global_dim(graphs_in),
        hidden_dim=HIDDEN_DIM,
        num_layers=int(depth),
        dropout=DROPOUT,
        residual=RESIDUAL,
        norm=NORM,
        target_dim=infer_target_dim(graphs_in),
    )


cfg = TrainConfig(
    lr=LR,
    batch_size=BATCH_SIZE,
    max_epochs=MAX_EPOCHS,
    patience=PATIENCE,
    num_workers=NUM_WORKERS,
    aux_losses=[WeightedLossTerm(
        name="edge",
        fn=edge_loss_term,
        weight=EDGE_LOSS_WEIGHT,
        params=EDGE_LOSS_PARAMS,
    )],
)
device = cfg.device

variants = {
    "original": {
        "label": "Original markers",
        "g_train": g_train,
        "g_val": g_val,
        "markers_removed": [],
        "seed_offset": 0,
    },
    REDUCED_VARIANT_KEY: {
        "label": REDUCED_VARIANT_LABEL,
        "g_train": g_train_reduced,
        "g_val": g_val_reduced,
        "markers_removed": markers_removed_for_reduced,
        "seed_offset": 1000,
    },
}

trained_models = {variant: {} for variant in variants}
training_rows = []

for variant_key, variant in variants.items():
    for depth in MODEL_DEPTHS:
        seed = SPLIT_SEED + variant["seed_offset"] + int(depth)
        set_all_seeds(seed)
        print(f"Training {variant['label']} | depth={depth}")
        model = make_gin_model(variant["g_train"], depth)
        model, metrics, history = train(
            model,
            variant["g_train"],
            variant["g_val"],
            cfg,
        )
        trained_models[variant_key][int(depth)] = model
        training_rows.append({
            "variant": variant_key,
            "variant_label": variant["label"],
            "depth": int(depth),
            "seed": int(seed),
            "val_mae_transformed": float(metrics["val_mae"]),
            "epochs_trained": len(history["val_mae"]),
        })
        if SAVE_MODEL_CHECKPOINTS:
            torch.save(
                model.state_dict(),
                MODELS_DIR / f"{variant_key}_depth{depth}.pt",
            )

training_summary_df = pd.DataFrame(training_rows)
training_summary_df.to_csv(TABLES_DIR / "training_summary.csv", index=False)
display(training_summary_df)
```

**Predict Validation Curvature**

```python
from src.inference.predict import predict_targets

prediction_rows = []
predictions = {}

for variant_key, variant in variants.items():
    predictions[variant_key] = {}
    for depth in MODEL_DEPTHS:
        y_true, y_pred, _ = predict_targets(
            variant["g_val"],
            trained_models[variant_key][int(depth)],
            device=device,
            batch_size=PREDICT_BATCH_SIZE,
            num_workers=NUM_WORKERS,
            target_transform=target_transform,
        )
        y_true = select_target_column(y_true, TARGET_INDEX_FOR_ANALYSIS)
        y_pred = select_target_column(y_pred, TARGET_INDEX_FOR_ANALYSIS)
        predictions[variant_key][int(depth)] = {
            "y_true": y_true,
            "y_pred": y_pred,
        }
        prediction_rows.append({
            "variant": variant_key,
            "depth": int(depth),
            "mse": float(np.nanmean(np.square(y_pred - y_true))),
            "mae": float(np.nanmean(np.abs(y_pred - y_true))),
        })

prediction_summary_df = pd.DataFrame(prediction_rows)
prediction_summary_df.to_csv(TABLES_DIR / "prediction_summary.csv", index=False)
display(prediction_summary_df)
```

**Mesh Prediction Illustration**

```python
from organograph.plotting.meshes import plot_organoid_mesh
from plotly.subplots import make_subplots

from src.plotting.mesh_plots import project_node_quantities_to_mesh


def find_exemplar_graph_index(graphs_in, keys, *, meta_lookup=None):
    keys = list(keys or [])
    for key in keys:
        for index, graph in enumerate(graphs_in):
            if graph_metadata_key(graph, meta_lookup=meta_lookup) == key:
                return index, key
    if not ALLOW_EXEMPLAR_FALLBACK:
        raise ValueError(f"None of the forced exemplar keys were found: {keys}")
    for index, graph in enumerate(graphs_in):
        metadata = meta_lookup.get(getattr(graph, "organoid_str", None), {})
        if metadata.get("mesh_path") is not None:
            return index, graph_metadata_key(graph, meta_lookup=meta_lookup)
    raise ValueError("No validation graph with a mesh_path was found.")


def zoomed_camera(view, distance):
    azimuth = np.deg2rad(float(view.get("azim", -135)))
    elevation = np.deg2rad(float(view.get("elev", 25)))
    distance = float(distance)
    return dict(
        eye=dict(
            x=float(distance * np.cos(elevation) * np.cos(azimuth)),
            y=float(distance * np.cos(elevation) * np.sin(azimuth)),
            z=float(distance * np.sin(elevation)),
        )
    )


def apply_scene_zoom(figure, scene_name):
    figure.layout[scene_name].update(
        camera=zoomed_camera(MESH_VIEW, MESH_CAMERA_DISTANCE),
        aspectmode="data",
    )


def add_subfigure_to_grid(figure, subfigure, row, col, scene_index):
    for trace in subfigure.data:
        figure.add_trace(trace, row=row, col=col)
    scene_name = "scene" if scene_index == 1 else f"scene{scene_index}"
    if subfigure.layout.scene is not None:
        figure.layout[scene_name].update(
            subfigure.layout.scene.to_plotly_json()
        )
    apply_scene_zoom(figure, scene_name)


def mesh_scalar_figure(vertex_values, *, show_colorbar=False):
    projection = project_node_quantities_to_mesh(
        selected_graph,
        node_true=true_values,
        node_pred=vertex_values,
        meta_lookup=val_meta_lookup,
    )
    return plot_organoid_mesh(
        projection["mesh"],
        vertex_values=projection["mesh_pred"],
        backend="plotly",
        colorscale=CURVATURE_COLORSCALE,
        center_at_zero=CENTER_CURVATURE_AT_ZERO,
        vmin=curvature_vmin,
        vmax=curvature_vmax,
        show_colorbar=show_colorbar,
        view=MESH_VIEW,
        fig_size=(MESH_PANEL_SIZE, MESH_PANEL_SIZE),
    )


def plot_variant_prediction_grid(variant_key, variant):
    figure = make_subplots(
        rows=2,
        cols=2,
        specs=[[{"type": "scene"}, {"type": "scene"}],
               [{"type": "scene"}, {"type": "scene"}]],
        subplot_titles=[
            "True curvature",
            f"Prediction | depth {MODEL_DEPTHS[0]}",
            f"Prediction | depth {MODEL_DEPTHS[1]}",
            f"Prediction | depth {MODEL_DEPTHS[2]}",
        ],
        horizontal_spacing=0.015,
        vertical_spacing=0.055,
    )

    panel_values = [
        true_values,
        prediction_values[variant_key][int(MODEL_DEPTHS[0])],
        prediction_values[variant_key][int(MODEL_DEPTHS[1])],
        prediction_values[variant_key][int(MODEL_DEPTHS[2])],
    ]
    panel_locations = [(1, 1), (1, 2), (2, 1), (2, 2)]

    for scene_index, (values, (row, col)) in enumerate(
        zip(panel_values, panel_locations),
        start=1,
    ):
        subfigure = mesh_scalar_figure(
            values,
            show_colorbar=(scene_index == 1),
        )
        add_subfigure_to_grid(figure, subfigure, row, col, scene_index)

    figure.update_layout(
        width=2 * MESH_PANEL_SIZE + 120,
        height=2 * MESH_PANEL_SIZE + 150,
        title=(
            f"{selected_key} | {variant['label']}"
            f"<br>shared scale [{curvature_vmin:.3g}, {curvature_vmax:.3g}]"
        ),
        margin=dict(l=20, r=20, t=95, b=20),
    )
    return figure


selected_graph_index, selected_key = find_exemplar_graph_index(
    g_val,
    split_info.get("forced_val_keys") or FORCED_VAL_KEYS,
    meta_lookup=val_meta_lookup,
)
selected_graph = g_val[selected_graph_index]
start, end = graph_slice_bounds(g_val, selected_graph_index)

true_values = predictions["original"][int(MODEL_DEPTHS[0])]["y_true"][start:end]
prediction_values = {
    variant_key: {
        int(depth): predictions[variant_key][int(depth)]["y_pred"][start:end]
        for depth in MODEL_DEPTHS
    }
    for variant_key in variants
}

all_values = [true_values]
for variant_key in variants:
    all_values.extend(
        prediction_values[variant_key][int(depth)]
        for depth in MODEL_DEPTHS
    )
all_numeric = np.concatenate([np.asarray(values).ravel() for values in all_values])
finite_numeric = all_numeric[np.isfinite(all_numeric)]
if CENTER_CURVATURE_AT_ZERO:
    curvature_limit = float(np.max(np.abs(finite_numeric)))
    curvature_limit = 1.0 if curvature_limit == 0 else curvature_limit
    curvature_vmin, curvature_vmax = -curvature_limit, curvature_limit
else:
    curvature_vmin = float(np.min(finite_numeric))
    curvature_vmax = float(np.max(finite_numeric))

mesh_figures = {}
for variant_key, variant in variants.items():
    figure = plot_variant_prediction_grid(variant_key, variant)
    mesh_figures[variant_key] = figure
    save_plotly_figure(figure, f"prediction_curvature_mesh_grid_{variant_key}")
    figure.show()

np.savez_compressed(
    ARRAYS_DIR / "exemplar_prediction_values.npz",
    graph_index=selected_graph_index,
    true_curvature=true_values,
    curvature_vmin=curvature_vmin,
    curvature_vmax=curvature_vmax,
    **{
        f"{variant_key}_pred_depth{depth}": prediction_values[variant_key][int(depth)]
        for variant_key in variants
        for depth in MODEL_DEPTHS
    },
)
```

**Export Mesh Fields**

```python
from organograph.mesh.OrganoidMesh import OrganoidMesh
import vtk
from vtk.util.numpy_support import numpy_to_vtk

mesh_export_projection = project_node_quantities_to_mesh(
    selected_graph,
    node_true=true_values,
    node_pred=true_values,
    meta_lookup=val_meta_lookup,
)
raw_mesh_path = Path(mesh_export_projection["mesh_path"])
raw_mesh = OrganoidMesh(str(raw_mesh_path))
plot_mesh = mesh_export_projection["mesh"]
vertex_owner = np.asarray(mesh_export_projection["vertex_owner"], dtype=np.int64)

np.savez_compressed(
    MESH_EXPORT_DIR / "raw_membrane_mesh.npz",
    vertices=np.asarray(raw_mesh.v, dtype=np.float32),
    faces=np.asarray(raw_mesh.f, dtype=np.int64),
    mesh_path=str(raw_mesh_path),
    selected_key=selected_key,
)
np.savez_compressed(
    MESH_EXPORT_DIR / "normalized_plot_mesh.npz",
    vertices=np.asarray(plot_mesh.v, dtype=np.float32),
    faces=np.asarray(plot_mesh.f, dtype=np.int64),
    vertex_owner=vertex_owner,
    mesh_path=str(raw_mesh_path),
    selected_key=selected_key,
)

mesh_curvature_fields = {
    "ground_truth": np.asarray(mesh_export_projection["mesh_true"], dtype=np.float32),
}
for variant_key in variants:
    for depth in MODEL_DEPTHS:
        projection = project_node_quantities_to_mesh(
            selected_graph,
            node_true=true_values,
            node_pred=prediction_values[variant_key][int(depth)],
            meta_lookup=val_meta_lookup,
        )
        if projection["mesh_pred"].shape[0] != vertex_owner.shape[0]:
            raise ValueError("Projected prediction field does not match exemplar mesh vertex count")
        mesh_curvature_fields[f"{variant_key}_pred_depth{int(depth)}"] = np.asarray(
            projection["mesh_pred"],
            dtype=np.float32,
        )

np.savez_compressed(
    MESH_EXPORT_DIR / "mesh_curvature_fields.npz",
    **mesh_curvature_fields,
    vertex_owner=vertex_owner,
    curvature_vmin=np.float32(curvature_vmin),
    curvature_vmax=np.float32(curvature_vmax),
    mesh_path=str(raw_mesh_path),
    selected_key=selected_key,
)

mesh_marker_labels = pd.Series(marker_projection["mesh_categories"]).astype("string").fillna("none")
marker_categories = sorted(mesh_marker_labels.unique().tolist())
marker_to_code = {marker: code for code, marker in enumerate(marker_categories)}
mesh_marker_codes = mesh_marker_labels.map(marker_to_code).to_numpy(dtype=np.int64)
marker_field_df = pd.DataFrame({
    "vertex_index": np.arange(len(mesh_marker_labels), dtype=np.int64),
    "cell_index": vertex_owner,
    "fate_marker": mesh_marker_labels.to_numpy(dtype=str),
    "fate_marker_code": mesh_marker_codes,
})
marker_field_df.to_csv(MESH_EXPORT_DIR / "mesh_fate_marker_field.csv", index=False)
pd.DataFrame({
    "fate_marker": marker_categories,
    "fate_marker_code": [marker_to_code[marker] for marker in marker_categories],
}).to_csv(MESH_EXPORT_DIR / "mesh_fate_marker_codebook.csv", index=False)
np.savez_compressed(
    MESH_EXPORT_DIR / "mesh_fate_marker_field.npz",
    fate_marker_code=mesh_marker_codes,
    fate_marker_categories=np.asarray(marker_categories, dtype=str),
    vertex_owner=vertex_owner,
    mesh_path=str(raw_mesh_path),
    selected_key=selected_key,
)

VTP_EXPORT_DIR = MESH_EXPORT_DIR / "vtp"
VTP_EXPORT_DIR.mkdir(parents=True, exist_ok=True)


def _read_raw_polydata(path):
    reader = vtk.vtkXMLPolyDataReader()
    reader.SetFileName(str(path))
    reader.Update()
    polydata = vtk.vtkPolyData()
    polydata.DeepCopy(reader.GetOutput())
    return polydata


def _add_numeric_point_array(polydata, name, values):
    values = np.asarray(values)
    n_points = int(polydata.GetNumberOfPoints())
    if values.shape[0] != n_points:
        raise ValueError(f"Point array {name!r} has length {values.shape[0]}, expected {n_points}")
    array = numpy_to_vtk(values.reshape(-1), deep=True)
    array.SetName(str(name))
    point_data = polydata.GetPointData()
    if point_data.HasArray(str(name)):
        point_data.RemoveArray(str(name))
    point_data.AddArray(array)


def _add_string_point_array(polydata, name, values):
    values = np.asarray(values, dtype=object).reshape(-1)
    n_points = int(polydata.GetNumberOfPoints())
    if values.shape[0] != n_points:
        raise ValueError(f"Point array {name!r} has length {values.shape[0]}, expected {n_points}")
    array = vtk.vtkStringArray()
    array.SetName(str(name))
    array.SetNumberOfValues(n_points)
    for index, value in enumerate(values):
        array.SetValue(index, str(value))
    point_data = polydata.GetPointData()
    if point_data.HasArray(str(name)):
        point_data.RemoveArray(str(name))
    point_data.AddArray(array)


def _write_vtp_with_point_fields(path, numeric_fields=None, string_fields=None):
    numeric_fields = {} if numeric_fields is None else dict(numeric_fields)
    string_fields = {} if string_fields is None else dict(string_fields)
    polydata = _read_raw_polydata(raw_mesh_path)
    n_points = int(polydata.GetNumberOfPoints())
    if n_points != vertex_owner.shape[0]:
        raise ValueError(f"Raw VTP has {n_points} points, expected {vertex_owner.shape[0]}")
    for name, values in numeric_fields.items():
        _add_numeric_point_array(polydata, name, values)
    for name, values in string_fields.items():
        _add_string_point_array(polydata, name, values)
    writer = vtk.vtkXMLPolyDataWriter()
    writer.SetFileName(str(path))
    writer.SetInputData(polydata)
    writer.Write()


base_vtp_numeric_fields = {
    "true_curvature": mesh_curvature_fields["ground_truth"],
    "fate_marker_code": mesh_marker_codes.astype(np.int32),
    "vertex_owner_cell_index": vertex_owner.astype(np.int32),
}
base_vtp_string_fields = {
    "fate_marker": mesh_marker_labels.to_numpy(dtype=str),
}
all_vtp_fields = dict(base_vtp_numeric_fields)
all_vtp_fields.update({
    name: values
    for name, values in mesh_curvature_fields.items()
    if name != "ground_truth"
})
all_fields_vtp = VTP_EXPORT_DIR / "exemplar_mesh_all_fields.vtp"
_write_vtp_with_point_fields(
    all_fields_vtp,
    numeric_fields=all_vtp_fields,
    string_fields=base_vtp_string_fields,
)

per_prediction_vtp_files = {}
ground_truth_vtp = VTP_EXPORT_DIR / "exemplar_mesh_ground_truth_and_markers.vtp"
_write_vtp_with_point_fields(
    ground_truth_vtp,
    numeric_fields=base_vtp_numeric_fields,
    string_fields=base_vtp_string_fields,
)
for field_name, field_values in mesh_curvature_fields.items():
    if field_name == "ground_truth":
        continue
    vtp_path = VTP_EXPORT_DIR / f"exemplar_mesh_{field_name}.vtp"
    _write_vtp_with_point_fields(
        vtp_path,
        numeric_fields={
            **base_vtp_numeric_fields,
            "predicted_curvature": field_values,
        },
        string_fields=base_vtp_string_fields,
    )
    per_prediction_vtp_files[field_name] = vtp_path.name


mesh_export_manifest = {
    "selected_key": selected_key,
    "selected_graph_index": selected_graph_index,
    "raw_mesh_path": raw_mesh_path,
    "export_dir": MESH_EXPORT_DIR,
    "vertex_count": int(vertex_owner.shape[0]),
    "face_count": int(np.asarray(raw_mesh.f).shape[0]),
    "curvature_fields": sorted(mesh_curvature_fields.keys()),
    "marker_categories": marker_categories,
    "files": {
        "raw_mesh": "raw_membrane_mesh.npz",
        "normalized_plot_mesh": "normalized_plot_mesh.npz",
        "curvature_fields": "mesh_curvature_fields.npz",
        "fate_marker_field_csv": "mesh_fate_marker_field.csv",
        "fate_marker_field_npz": "mesh_fate_marker_field.npz",
        "fate_marker_codebook": "mesh_fate_marker_codebook.csv",
        "vtp_dir": "vtp",
        "vtp_all_fields": f"vtp/{all_fields_vtp.name}",
        "vtp_ground_truth_and_markers": f"vtp/{ground_truth_vtp.name}",
        "vtp_per_prediction": {
            field_name: f"vtp/{filename}"
            for field_name, filename in per_prediction_vtp_files.items()
        },
    },
}
with open(MESH_EXPORT_DIR / "mesh_field_manifest.json", "w") as handle:
    json.dump(jsonable(mesh_export_manifest), handle, indent=2)

print(f"Saved mesh-aligned exemplar fields to {MESH_EXPORT_DIR}")
```
